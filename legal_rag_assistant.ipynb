```python
from datasets import load_dataset

dataset = load_dataset(
    "theatticusproject/cuad",
    split="train[:50]",
    verification_mode="no_checks"
)

print("Dataset size:", len(dataset))
print("Columns:", dataset[0].keys())
```

```python
documents = []

for idx, item in enumerate(dataset):
    pdf_obj = item["pdf"]
    text_parts = []

    try:
        for page in pdf_obj.pages:
            text = page.extract_text()
            if text:
                text_parts.append(text.strip())
    except Exception:
        continue

    full_text = "\n".join(text_parts)

    if full_text.strip():
        documents.append({
            "text": full_text,
            "metadata": {
                "doc_id": idx
            }
        })

print("Documents extracted:", len(documents))
print(documents[0]["text"][:500] if documents else "No documents extracted")
```

```python
filtered_documents = []

for doc in documents:
    first_part = doc["text"][:500].lower()
    if "datasheet for contract understanding atticus dataset" in first_part:
        continue
    filtered_documents.append(doc)

documents = filtered_documents
print("Filtered documents:", len(documents))
```

```python
def chunk_text(text, chunk_size=1000, overlap=200):
    chunks = []
    start = 0
    text = " ".join(text.split())

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap

    return chunks


chunked_docs = []

for doc in documents:
    chunks = chunk_text(doc["text"])
    for i, chunk in enumerate(chunks):
        chunked_docs.append({
            "text": chunk,
            "metadata": {
                **doc.get("metadata", {}),
                "chunk_id": i
            }
        })

print("Chunks created:", len(chunked_docs))
print(chunked_docs[0]["text"][:400] if chunked_docs else "No chunks created")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

texts = [doc["text"] for doc in chunked_docs]
embeddings = embedding_model.encode(texts, normalize_embeddings=True)

print("Embeddings created:", len(embeddings))
```

```python
import faiss
import numpy as np

embedding_matrix = np.array(embeddings).astype("float32")
dimension = embedding_matrix.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(embedding_matrix)

print("FAISS index built with", index.ntotal, "vectors")
```

```python
def retrieve(query, top_k=5):
    query_embedding = embedding_model.encode([query], normalize_embeddings=True)
    query_embedding = np.array(query_embedding).astype("float32")

    distances, indices = index.search(query_embedding, top_k)

    results = []
    for idx, dist in zip(indices[0], distances[0]):
        results.append({
            "text": chunked_docs[idx]["text"],
            "metadata": chunked_docs[idx].get("metadata", {}),
            "distance": float(dist)
        })
    return results
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
llm = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
llm = llm.to(device)

print("Loaded on:", device)
```

```python
def generate_answer(query, context, max_new_tokens=80):
    prompt = f"""
Answer the question using ONLY the context.

If the answer is yes/no, explain WHY using the context.

If the answer is not found, say: I don't know.

Context:
{context}

Question:
{query}

Answer:
""".strip()

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    ).to(device)

    outputs = llm.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
def ask(query, top_k=5):
    results = retrieve(query, top_k=top_k)

    context = "\n\n".join([r["text"][:350] for r in results])
    answer = generate_answer(query, context)

    sources = []
    for i, r in enumerate(results, 1):
        sources.append(
            f"""[Chunk {i}] distance={r['distance']:.4f} | doc_id={r['metadata'].get('doc_id')} | chunk_id={r['metadata'].get('chunk_id')}
{r['text'][:400]}"""
        )

    return {
        "question": query,
        "answer": answer,
        "sources": "\n\n".join(sources)
    }
```

```python
result = ask("What disclaimer or limitation clauses are mentioned?", top_k=5)

print("Question:", result["question"])
print("\nAnswer:\n", result["answer"])
print("\nSources:\n", result["sources"][:1500])
```

```python
def format_answer_text(answer):
    cleaned = answer.replace("[", "").replace("]", "")
    parts = [p.strip() for p in cleaned.split(",")]
    parts = [p for p in parts if p]

    if len(parts) <= 1:
        return answer

    return "\n".join([f"• {p}" for p in parts])
```

```python
def hallucination_score(answer, results):
    context = " ".join([r["text"] for r in results]).lower()
    answer_words = set(answer.lower().split())

    overlap = len(answer_words & set(context.split()))
    score = overlap / max(len(answer_words), 1)

    return score
```

```python
def rag_ui(query, top_k):
    if not query.strip():
        return "Please enter a question.", ""

    results = retrieve(query, top_k=int(top_k))
    context = "\n\n".join([r["text"][:350] for r in results])
    answer = generate_answer(query, context)

    score = hallucination_score(answer, results)

    if score < 0.3:
        confidence = "⚠️ Low confidence"
    elif score < 0.6:
        confidence = "⚠️ Medium confidence"
    else:
        confidence = "✅ High confidence"

    final_answer = f"{answer}\n\nConfidence: {confidence}\n\n(Answer based on retrieved contract clauses)"

    return final_answer, sources
```

```python
import gradio as gr

sample_questions = [
    "What disclaimer or limitation clauses are mentioned?",
    "Are there indemnification clauses?",
    "Are there third-party beneficiary clauses?",
    "Are there warranty disclaimer clauses?",
    "Are there governing law or dispute clauses?"
]

with gr.Blocks(title="Legal Document RAG Assistant") as demo:
    gr.Markdown("# Legal Document RAG Assistant")
    gr.Markdown("Ask questions about CUAD contract documents using RAG + FAISS + FLAN-T5.")

    with gr.Row():
        query = gr.Textbox(
            label="Your Question",
            placeholder="Type a legal contract question..."
        )
        top_k = gr.Slider(
            minimum=1,
            maximum=10,
            value=5,
            step=1,
            label="Top-K Retrieval"
        )

    ask_btn = gr.Button("Ask")

    answer = gr.Textbox(label="Answer", lines=6)
    sources = gr.Textbox(label="Retrieved Sources", lines=18)

    ask_btn.click(
        fn=rag_ui,
        inputs=[query, top_k],
        outputs=[answer, sources]
    )

    gr.Examples(
        examples=[[q, 5] for q in sample_questions],
        inputs=[query, top_k]
    )

demo.launch(share=True)
```

```python
test_cases = [
    {
        "question": "Are there indemnification clauses?",
        "expected_keywords": ["indemnification"]
    },
    {
        "question": "Are there warranty disclaimer clauses?",
        "expected_keywords": ["warranty", "disclaimer"]
    },
    {
        "question": "Are there limitation of liability clauses?",
        "expected_keywords": ["limitation", "liability"]
    }
]

def evaluate():
    retrieval_hits = 0
    answer_hits = 0
    halluc_scores = []

    for test in test_cases:
        print("=" * 60)
        question = test["question"]
        expected = test["expected_keywords"]

        # Retrieval
        results = retrieve(question, top_k=5)

        retrieved_text = " ".join([r["text"] for r in results]).lower()
        retrieval_match = any(k in retrieved_text for k in expected)

        # Answer
        context = "\n\n".join([r["text"][:350] for r in results])
        answer = generate_answer(question, context)

        answer_match = any(k in answer.lower() for k in expected)

        # Hallucination
        score = hallucination_score(answer, results)
        halluc_scores.append(score)

        print("Question:", question)
        print("Answer:", answer)
        print("Retrieval Match:", retrieval_match)
        print("Answer Match:", answer_match)
        print("Hallucination Score:", round(score, 2))

        if retrieval_match:
            retrieval_hits += 1
        if answer_match:
            answer_hits += 1

    total = len(test_cases)

    print("\nFINAL METRICS:")
    print("Retrieval Accuracy (Hit@K):", retrieval_hits / total)
    print("Answer Accuracy:", answer_hits / total)
    print("Avg Hallucination Score:", sum(halluc_scores)/total)
```

```python
query = "Are there indemnification clauses?"

# Step 1: retrieve
results = retrieve(query, top_k=5)

# Step 2: generate answer
context = "\n\n".join([r["text"][:350] for r in results])
answer = generate_answer(query, context)

# Step 3: hallucination score
score = hallucination_score(answer, results)

print("Answer:", answer)
print("Hallucination Score:", round(score, 2))
```
